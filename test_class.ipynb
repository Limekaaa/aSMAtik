```python
from mesa import Agent, Model
from src import agents
from mesa.space import MultiGrid
from mesa.datacollection import DataCollector
from itertools import product
```

```python
class MoneyModel(Model):
    """A model with some number of agents."""
    def __init__(self, N=100, width=10, height=10, seed=None):
        """Initialize the model.

        Args:
            n (int, optional): Number of agents. Defaults to 100.
            width (int, optional): Grid width. Defaults to 10.
            height (int, optional): Grid height. Defaults to 10.
            seed (int, optional): Random seed. Defaults to None.
        """

        #TO COMPLETE
        super().__init__(seed=seed)
        self.grid = MultiGrid(width, height, torus=True)
        self.datacollector = DataCollector(agent_reporters={"Wealth": "wealth"})

        agents_coords = self.random.choices(list(product([i for i in range(width)], [i for i in range(height)])), k=N)

        for i in range(N):
            #new_agent = MoneyAgent(self)    
            new_agent = agents.greenAgent(model=self, policy_name="random_policy")
            self.grid.place_agent(new_agent, agents_coords[i])

    def step(self):
        #TO COMPLETE
        self.agents.shuffle_do("step_agent")

    def get_all_agents_wealth(self):
      #To complete
      #return [agent.wealth for agent in self.agents]
      self.datacollector.collect(self)
      return self.datacollector.get_agent_vars_dataframe()
```

```python
model = MoneyModel(30, 10, 10)
for _ in range(100):
    model.step()
display(model.get_all_agents_wealth())
```
